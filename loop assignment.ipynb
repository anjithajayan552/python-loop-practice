8. Read a number and find the sum of its digits. Example: 4521 gives 12.

```python
num = int(input("enter a number:"))
total = 0
while num > 0:
  digit = num % 10
  total = total + digit
  num = num // 10
print("sum:", total)
```

9. Read a number, reverse it using a while loop, and check whether it is a palindrome.
Example: 121 reversed is 121.

```python
num = int(input("enter a number:"))
original = num
reverse = 0
while num > 0:
  digit = num % 10
  reverse = reverse * 10 + digit
  num = num // 10
print("reverse :", reverse)
if original == reverse:
  print("palindrome")
else:
  print("not pallindrome")
```

10. Check whether a 3-digit number is an Armstrong number. Example: 153 = 1³ + 5³ + 3³.

```python
n = int(input("enter the number:"))
original = n
total = 0
while n> 0:
  digit = n % 10
  total = total + digit ** 3
  n = n //10
if total == original:
  print("Amstrong number")
else:
  print(" not an Amstorng number")
```

11. Monthly savings. Rahul saves ₹500 in the first month and increases the deposit by
₹100 every month. In how many months will his total savings reach ₹10,000?

```python
deposit = 500
total = 0
month = 0
while total < 10000:
  total = total + deposit
  month = month +1
  deposit = deposit + 100
print("months required:", month)
print("total savings:", total)
```

12. Shopping bill. Keep reading item prices until the user enters 0. Print the total. If the
total is above ₹1000, give a 10% discount and print the final amount to pay.

```python
total = 0
while True:
  price = float (input("enter the item price( o to stop):"))

  if price == 0:
    break
  total = total + price
print("total:", total)

if total >1000:
  discount = total * 0.10
  total = total - discount
print("final amount to pay:", total)

```

13. List. A teacher stores the marks of five students in the list marks = [45, 78, 32, 90, 66].
Using a while loop, print each student&#39;s mark with the student number, then print the
total, the average and the highest mark.

```python
marks = [45, 78, 32, 90, 66]
i = 0
total = 0
highest = marks[0]
while i < len(marks):
    print("Student", i + 1, ":", marks[i])
    total = total + marks[i]
    if marks[i] > highest:
        highest = marks[i]
    i = i + 1
average = total / len(marks)
print("Total =", total)
print("Average =", average)
print("Highest mark =", highest)
```

14. Tuple. A shop stores product prices in the tuple prices = (120, 45, 300, 80, 150). Read
a price from the user and use a while loop to search for it. Print its position in the tuple,
or &quot;Not found&quot;.


```python
prices = (120, 45, 300, 80, 150)
price = int(input("Enter a price to search: "))
i = 0
found = False
while i < len(prices):
    if prices[i] == price:
        print("Found at position", i + 1)
        found = True
        break
    i = i + 1
if found == False:
    print("Not found")
```
